# **4. Data Cleansing-Standarisasi kolom Kode Cafe**

Algoritma yang akan digunakan:
1. Cek apakah kode pelanggan memiliki format "TXN-" di awal.
2. Cek jumlah digit setelah "TXN-". Jika hanya terdapat 4 digit, tambahkan satu
   digit nol di depan sehingga menjadi 7 digit.
3. Tampilkan hasil yang sudah diperbaiki.

```python
# Function to format 'Transaction ID'
def format_transaction_id(transaction_id):
    if isinstance(transaction_id, str):
        if transaction_id.startswith("TXN_"):
            angka = transaction_id.split('_')[-1]
            if angka.isdigit():
                return "TXN_" + angka.zfill(7)
    return transaction_id
```

```python
# Apply the function to the 'Transaction ID' column
data['Transaction ID'] = data['Transaction ID'].apply(format_transaction_id)
```

```python
# Display the updated 'Transaction ID' column
data[['Transaction ID']].head()
```

Kolom Kode Transaction telah diperbaiki. Tidak ada kode pelanggan dengan 4 digit yang perlu ditambahkan angka nol, karena semua sudah memiliki 7 digit setelah "TXN-".

# **5. Data Cleansing-Standarisasi kolom Transaction ID**

```python
import pandas as pd
import re

# Baca file .csv
file_path = '/content/drive/MyDrive/Colab Notebooks/DatasetCafe/dirty_cafe_sales.csv'
df = pd.read_csv(file_path)

# Define a function to clean the 'Transaction ID' column
def clean_transaction_id(transaction_id):
    # Define patterns to remove
    patterns = [
        r'\s+',          # Menghapus spasi
        r'[^A-Z0-9_]',   # Menghapus karakter selain huruf, angka, dan _
    ]

    # Compile the patterns into a single regex pattern
    combined_pattern = re.compile('|'.join(patterns), flags=re.IGNORECASE)

    # Remove the patterns
    cleaned_transaction_id = combined_pattern.sub('', str(transaction_id)).strip().upper()

    return cleaned_transaction_id

# Apply the cleaning function to the 'Transaction ID' column
df['Transaction ID Bersih'] = df['Transaction ID'].apply(clean_transaction_id)

# Display the cleaned DataFrame
df[['Transaction ID', 'Transaction ID Bersih']].head()
```

# **6. Data Cleansing-Standarisasi kolom Transaction Date**

```python
# Define a function to standardize the date format
def standardize_date(date):
    try:
        # Convert the date to datetime format and then to the desired format
        standardized_date = pd.to_datetime(date, errors='coerce').strftime('%d-%m-%Y')
    except:
        standardized_date = None
    return standardized_date

# Apply the standardization function to the 'Transaction Date' column
df['Transaction Date Standar'] = df['Transaction Date'].apply(standardize_date)

# Display the DataFrame with the standardized dates
df[['Transaction Date', 'Transaction Date Standar']].head()
```

# **7. Data Cleansing-Deduplikasi Data**

Menampilkan data yang sama dalam semua kolom

```python
# Find duplicate rows based on columns 'Item', 'Quantity', 'Price Per Unit',
# 'Total Spent', 'Payment Method', 'Location', and 'Transaction Date'
duplicate_rows = df[df.duplicated([
    'Item',
    'Quantity',
    'Price Per Unit',
    'Total Spent',
    'Payment Method',
    'Location',
    'Transaction Date'
], keep=False)]

# Display the duplicate rows
duplicate_rows
```

Data duplikat berdasarkan semua kolom

```python
# Sort the dataframe by 'Transaction ID' to ensure that the smallest Transaction ID is kept
df_sorted = df.sort_values(by='Transaction ID')

# Drop duplicates based on the transaction details, keeping the first (smallest Transaction ID)
df_deduplicated = df_sorted.drop_duplicates(
    subset=[
        'Item',
        'Quantity',
        'Price Per Unit',
        'Total Spent',
        'Payment Method',
        'Location',
        'Transaction Date'
    ],
    keep='first'
)

# Display the resulting dataframe
df_deduplicated.reset_index(drop=True, inplace=True)
df_deduplicated.head()
```
